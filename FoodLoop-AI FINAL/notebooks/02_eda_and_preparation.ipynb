```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split

df = pd.read_csv("../data/food_demand.csv")
df["date"] = pd.to_datetime(df["date"])
df.head()
```

```python
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="expected_attendance", y="meals_consumed")
plt.title("Expected Attendance vs Meals Consumed")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="is_weekend", y="meals_consumed")
plt.title("Meals Consumed: Weekday vs Weekend")
plt.xlabel("0 = Weekday, 1 = Weekend")
plt.show()
```

```python
features = [
    "day_of_week", "is_weekend", "exam_week", "special_event",
    "temperature_c", "menu_type", "expected_attendance"
]

X = df[features]
y = df["meals_consumed"]

print("X:", X.shape)
print("y:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
import pandas as pd
import joblib

model = joblib.load("../models/food_demand_model.joblib")
```

```python
new_day = pd.DataFrame({
    "day_of_week": ["Monday"],
    "is_weekend": [0],
    "exam_week": [0],
    "special_event": [0],
    "temperature_c": [28.0],
    "menu_type": ["North Indian"],
    "expected_attendance": [430]
})
new_day
```

```python
predicted_demand = round(float(model.predict(new_day)[0]))
safety_buffer = max(5, round(predicted_demand * 0.025))
recommended_meals = predicted_demand + safety_buffer

print("Predicted demand:", predicted_demand)
print("Safety buffer:", safety_buffer)
print("Recommended preparation:", recommended_meals)
```

```python
traditional_preparation = 500
traditional_leftover = max(traditional_preparation - predicted_demand, 0)
ai_leftover = max(recommended_meals - predicted_demand, 0)
meals_saved = max(traditional_leftover - ai_leftover, 0)

print("Traditional leftovers:", traditional_leftover)
print("AI leftovers:", ai_leftover)
print("Potential overproduction avoided:", meals_saved)
```

```python
average_meal_weight_kg = 0.35
cost_per_meal_rupees = 45

food_saved_kg = meals_saved * average_meal_weight_kg
cost_saved = meals_saved * cost_per_meal_rupees

print(f"Food potentially saved: {food_saved_kg:.2f} kg")
print(f"Preparation cost potentially avoided: ₹{cost_saved:,.0f}")
```

```python
if meals_saved > 20:
    recommendation = "Reduce future preparation slightly and review the recurring waste pattern."
elif meals_saved > 0:
    recommendation = "A small amount of overproduction was avoided. Keep monitoring."
else:
    recommendation = "Keep the current recommendation and monitor for shortages."

print("FoodLoop AI:", recommendation)
```
